```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("titanic.csv")

#RATA-RATA usia, fare, Survive, pclass
print("\n===== Rata-Rata =====")
mean_age = df['age'].mean()
mean_fare = df['fare'].mean()
mean_survived = df['survived'].mean() * 100
mean_pclass = df['pclass'].mean()

print(f"Rata-rata Usia: {mean_age:.2f}tahun")
print(f"Rata-rata fare: {mean_fare:.2f}")
print(f"Persentase survive: {mean_survived:.2f}%")
print(f"Rata-rata pclass: {mean_pclass:.2f}")

print("\n===== Persentase Pclass =====")
class_percent = df['pclass'].value_counts(normalize=True) * 100
print(class_percent)

#Max/Min Usia dan Fare
max_usia = df['age'].max()
min_usia = df['age'].min()

max_fare = df['fare'].max()
min_fare = df['fare'].min()

print("=== Max & Min ===")
print(f"USIA Tertua: {max_usia}")
print(f"USIA Termuda: {min_usia}")
print(f"Fare Termahal: {max_fare}")
print(f"Fare Termurah: {min_fare}")

#IQR
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 *  IQR
upper_bound = Q3 + 1.5 * IQR

#IDENTIFIKASI OUTLIER
outliers = df[(df['fare'] < lower_bound) | (df['fare'] > upper_bound) * 100]
percent_outlier = (len(outliers)/len(df)) * 100

print("===== Outlier Fare Record =====")
print(f"Jumlah outlier fare: {len(outliers)}record")
print(f"Persentase outlier fare: {percent_outlier:.2f}%")


plt.figure(figsize=(8,4))
plt.boxplot(df['fare'])
plt.title("Outlier Fare (Boxplot)")
plt.xlabel("fare")
plt.show()

plt.figure(figsize=(8,4))
plt.hist(df['fare'], bins=50)
plt.title("Distribusi Fare")
plt.xlabel("fare")
plt.ylabel("Frekuensi")
plt.show()
```
